```python
import os
import shutil
```

```python

def convert_pth_to_pkl(pth_file_path):
    """Renames a .pth file to .pkl."""
    if not os.path.exists(pth_file_path):
        raise FileNotFoundError(f"File not found: {pth_file_path}")
    
    if not pth_file_path.lower().endswith('.pth'):
        raise ValueError("Input file must be a .pth file.")
    
    pkl_path = pth_file_path[:-4] + '.pkl' 
    shutil.move(pth_file_path, pkl_path)
    return pkl_path

```

```python
val = input("Enter your value: ")
new_path = convert_pth_to_pkl(val)
print("The file was renamed")
```
